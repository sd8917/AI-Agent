## Streaming
- Most models can stream their output content while it is being generated. By displaying output progressively, streaming significantly improves user experience,
particularly for longer responses.Calling stream() returns an iterator that yields output chunks as they are produced. You can use a loop to process each chunk in real-time.

```python
model.invoke("Write a 10 words paragraph on Artificial intelligent in aiops")
```

```python
for chunk in model.stream("Why do parrots have colorful feathers?"):
    print(chunk.text, end="|", flush=True)
```
